```python
tables = ['brands', 'categories', 'customers', 'order_items',
          'orders', 'products', 'staffs', 'stocks', 'stores']

for t in tables:
    df = spark.read.csv(f'{resource_path}{t}.csv',
                        header=True, inferSchema=True, sep=',')
    (df.write
       .mode('overwrite')
       .format('delta')
       .option('overwriteSchema', 'true')
       .save(f'{bronze_path}{t}'))
    spark.sql(f"""
        CREATE TABLE IF NOT EXISTS bikestore.logistics.bronze_{t}
        LOCATION '{bronze_path}{t}/'
    """)
    print(f'{t}: {df.count()} linhas')
```
